```python
import pygame
pygame.init()

# Criando uma janela
screen = pygame.display.set_mode((400, 300))
pygame.display.set_caption("Jogo Básico")

# Cor de fundo
screen.fill((255, 0, 0))

# Atualizando a tela
pygame.display.update()

# Esperar um tempo e fechar
pygame.time.wait(3000)
pygame.quit()
```
